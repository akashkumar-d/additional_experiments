# 33 configurations, repeated runs, variance estimates

**AC request:** *"Show all 33 configurations. Present results (or at least summary statistics) across all 33 student-teacher configurations, including weak or negative cases, with repeated runs and variance estimates. Explain why three configurations are omitted from Figure 3D."*

**What this notebook does:**
1. Defines the **exact 33-config grid** from the submission: 9 ReLU (`relu_plateau_grid.py`) + 12 round-1 GELU/SiLU (`smooth_teacher_grid.py`) + 12 round-2 large-$r_t$ GELU/SiLU (`smooth_plateau_v2.py`). All $p = 100$, $n = 15{,}000$, full-batch, $T = 8000$ steps.
2. Reruns **all 33 × 5 seeds** $\{5, 11, 23, 37, 41\}$ = 165 runs with the patched runner (per-step dense loss for $\rho_2$; polar/leakage metrics for $\tau_V^2$). Restartable.
3. Produces the **complete summary-statistics table**: per config, median ± IQR across seeds for every headline metric — including weak/divergent cases, no filtering.
4. Regenerates the **corrected Figure 3(d)** with the window-free time-to-target drift estimator, **all 33 configs included** (the original panel's slope estimator silently dropped 3 saturated configs — see §5 below), with seed error bars.
5. Emits `REPORT.md` with rebuttal-ready tables and the Fig-3D explanation.

**Why 3 configs were omitted from Fig 3(d) (the answer we will give):** the panel's drift estimator was the late-half secant slope with a filter that skips configs whose $\cos^2_{\min}$ changes by $< 0.005$ in the late half. The three dropped configs (ReLU $r_t{=}4\,\eta{=}2.0$, GELU $r_t{=}4\,\eta{=}0.7$, SiLU $r_t{=}4\,\eta{=}0.85$) are exactly the ones that align **fastest** — their $\cos^2_{\min}$ saturates near $1$ before the measurement window opens, so the estimator misreads them as 'no growth'. The corrected estimator (time to traverse a fixed alignment band) is window-free and includes them.

**Wall clock.** 165 runs × 8000 steps ≈ 100–120 h single-core. The run cell is restartable — run in chunks, or split the `ALL_RUNS` list across machines by slicing.

## 2. Run — multi-machine sharding + within-node parallelism

**To split across several notebook machines (e.g. multiple A100 links):**
1. Put an identical copy of this notebook + the `shared/` folder on every machine.
2. On each machine set a different `WORKER_ID` in the cell below (0, 1, 2, ...) and the same `NUM_WORKERS` (= number of machines).
3. Run the parallel run cell. Each machine computes a disjoint, load-balanced (round-robin) shard of the 165 runs and writes its own `item1_all33/*.npz`.
4. When all machines finish, download every machine's `item1_all33/` folder and **merge the `.npz` files into one folder** (filenames are globally unique — config + seed — so merging is just copying). Then run the analysis cells (§3 onward) on the merged folder.

**Within-node parallelism:** these are NumPy/CPU jobs (the GPU is not used), so the parallel cell also runs `N_PROC` worker processes on the node's CPU cores. On a typical A100 node with 32–128 cores, `N_PROC = 8–16` is safe. Wall-clock estimate: `110 h / (NUM_WORKERS x N_PROC)` — e.g. 4 machines x 8 processes ≈ **3.5 h**.

Both cells are restartable: completed checkpoints are skipped on rerun.

```python
# ---- EDIT THESE PER MACHINE ----
WORKER_ID   = 0        # 0 .. NUM_WORKERS-1, unique per machine
NUM_WORKERS = 1        # total number of machines running this notebook
USE_GPU     = False     # torch float64 on the GPU (recommended on A100 links)
DEVICES     = ['cuda']   # multi-GPU node: ['cuda:0','cuda:1','cuda:2','cuda:3']   # 'cuda' | 'cuda:0' | 'cpu'
N_PROC      = 16        # worker processes: GPU -> 3-4; CPU-only -> cores // 4
# --------------------------------
# NOTE (GPU): do NOT import torch in this notebook before launching the
# pool -- the children import it after fork (fork + initialized CUDA breaks).

MY_RUNS = ALL_RUNS[WORKER_ID::NUM_WORKERS]
print(f'Worker {WORKER_ID}/{NUM_WORKERS}: {len(MY_RUNS)} of {len(ALL_RUNS)} runs, '
      f'N_PROC={N_PROC}, backend={"torch:"+",".join(DEVICES) if USE_GPU else "numpy"}')
print('First few:', [(g, s) for (g, *_, s) in MY_RUNS[:6]])
```

### 2a. Parallel run (recommended) — uses N_PROC processes on this machine

```python
import multiprocessing as mp
import concurrent.futures as cf
from parallel_worker import run_full, run_full_gpu

if USE_GPU:
    worker_fn = run_full_gpu
    ARGS = [(str(OUT), N_STEPS, LOG_EVERY, gid, ta, p, rt, rs, n, eta, seed,
             DEVICES[i % len(DEVICES)])
            for i, (gid, ta, p, rt, rs, n, eta, seed) in enumerate(MY_RUNS)]
else:
    worker_fn = run_full
    ARGS = [(str(OUT), N_STEPS, LOG_EVERY, gid, ta, p, rt, rs, n, eta, seed)
            for (gid, ta, p, rt, rs, n, eta, seed) in MY_RUNS]

t0 = time.time(); done = 0
with cf.ProcessPoolExecutor(max_workers=N_PROC,
                            mp_context=mp.get_context('fork')) as ex:
    for gid, seed in ex.map(worker_fn, ARGS):
        done += 1
        el = time.time() - t0
        rem = el / done * (len(ARGS) - done)
        print(f'[{done}/{len(ARGS)}] {gid} seed={seed} done  '
              f'({el/3600:.2f} h elapsed, ~{rem/3600:.2f} h left)', flush=True)
print(f'Worker {WORKER_ID} finished in {(time.time()-t0)/3600:.2f} h')
```

### 2b. Serial fallback (single process; same shard; restartable)

```python
TOTAL_BUDGET_S = 400000   # ~110 h; rerun cell to resume after interruption
CHUNK_STEPS    = 4000
RUNS = MY_RUNS            # shard from the sharding cell above

t_start = time.time()
for i, (gid, ta, p, rt, rs, n, eta, seed) in enumerate(RUNS):
    elapsed = time.time() - t_start
    if elapsed > TOTAL_BUDGET_S:
        print(f'\n[budget] stop after {i}/{len(RUNS)}')
        break
    name = f'{ta}_p{p}_rt{rt}_rs{rs}_n{n}_eta{eta}_seed{seed}.npz'
    ckpt = OUT / name
    print(f'\n[{i+1}/{len(RUNS)}] {gid} seed={seed}: {name}')
    run_one(ta, p, rt, rs, n, eta, seed=seed, n_steps=N_STEPS,
            log_every=LOG_EVERY, ckpt_path=ckpt,
            chunk_steps=CHUNK_STEPS,
            budget_s=max(60.0, TOTAL_BUDGET_S - elapsed))
print(f'\n[done] {time.time()-t_start:.0f}s elapsed')
```

## 3. Per-run metric extraction

For every checkpoint: $\rho_2$ (late-half, per-step dense loss), cycle-mean plateau loss and oracle ratio, late-half $\Delta\cos^2_{\min}$, end alignment, $\tau_V^2$, drift time-to-target $t_{0.5}$ and band slope, init alignment, divergence flag. **No config is filtered out** — weak and divergent runs are tabulated alongside the clean ones.

```python
def _dedupe(steps, ys):
    s = np.asarray(steps, dtype=float); y = np.asarray(ys, dtype=float)
    if len(s) < 2: return s, y
    keep = np.concatenate([[True], np.diff(s) > 0])
    return s[keep], y[keep]

def smooth(x, k=11):
    x = np.asarray(x, dtype=float)
    if len(x) < k or k < 2: return x
    pad = k // 2
    xp = np.concatenate([np.full(pad, x[0]), x, np.full(k-1-pad, x[-1])])
    return np.convolve(xp, np.ones(k)/k, mode='valid')

def rho2_late(z):
    if 'L_full_dense' not in z.files or len(z['L_full_dense']) < 200: return float('nan')
    L = np.asarray(z['L_full_dense'], dtype=float); L = L[len(L)//2:]
    dL = np.diff(L); a, b = dL[:-1], dL[1:]
    da, db = a - a.mean(), b - b.mean()
    den = math.sqrt(float(np.dot(da,da)) * float(np.dot(db,db))) + 1e-30
    return float(-np.dot(da, db) / den)

def t_to(s, y, target, k=11):
    ys = smooth(y, k)
    idx = np.where(ys >= target)[0]
    return float(s[int(idx[0])]) if len(idx) else float('nan')

def band_slope(s, y, lo=0.05, hi=0.5, k=11):
    t_lo, t_hi = t_to(s, y, lo, k), t_to(s, y, hi, k)
    if math.isnan(t_lo) or math.isnan(t_hi) or t_hi <= t_lo: return float('nan')
    return (hi - lo) / (t_hi - t_lo)

def plateau_onset(L, win=500, tol=0.02):
    """First step from which the cycle-mean loss never again falls faster than
    `tol` per `win` steps. This is where the plateau begins, per run.

    Replaces the old fixed late-half window. A fixed window is stringent in the
    wrong way: it demands growth in [T/2, T] regardless of when the plateau
    actually formed, so a configuration that plateaus at step 300 and learns
    from 300 to 3000 scores zero, while the interval where the loss really was
    stalled is ignored. Measuring between the plateau onset and T asks the
    question the paper actually makes: do features improve WHILE the loss is
    stalled?
    """
    Lp = 0.5 * (L[:-1] + L[1:])                # cycle mean, parity removed
    T = len(Lp)
    if T <= win + 5:
        return None
    ts = np.arange(win, T)
    ok = (Lp[ts] - Lp[ts - win]) / np.maximum(Lp[ts - win], 1e-30) >= -tol
    idx, run_ok = None, True
    for i in range(len(ok) - 1, -1, -1):
        if ok[i] and run_ok: idx = i
        else: run_ok = False
    return int(ts[idx]) if idx is not None else None


def sat_onset(s, y, frac=0.99, floor=0.05):
    """First logged step at which mean cos^2 reaches `frac` of its FINAL value,
    i.e. the step by which feature learning is effectively finished.

    Why this is needed alongside plateau_onset. The plateau window only carries
    information when the features are still moving at the moment the loss
    stalls. Some configurations align so fast that they finish long before the
    loss flattens -- S1 reaches cos2=0.985 by step 280 while its loss goes on
    falling another 22% until step ~2745. For those the plateau-window gain is
    ~0 BY CONSTRUCTION, and reading that as "no decoupling" is wrong: the
    decoupling is simply carried by a different observable. It shows up not as
    "features improve while the loss is flat" but as "features are finished and
    the loss STILL sits far above what those features permit".

    We deliberately do NOT loosen plateau_onset to make these configs score.
    Declaring a plateau while the loss is still dropping 22-35% would be
    indefensible; the onset test stays strict and we measure the right quantity
    instead.
    """
    y = np.asarray(y, dtype=float)
    if len(y) < 2 or y[-1] < floor:
        return float('nan')
    idx = np.where(y >= frac * y[-1])[0]
    return float(s[int(idx[0])]) if len(idx) else float('nan')


def extract(ckpt):
    try:
        z = np.load(ckpt, allow_pickle=True); _ = [z[k] for k in z.files]
    except Exception as e:
        print(f'[skip corrupt] {ckpt.name}: {type(e).__name__}'); return None
    s, y = _dedupe(z['step'], z['cos2_min_AGOP'])
    L_dense = np.asarray(z['L_full_dense'], dtype=float) if 'L_full_dense' in z.files else np.asarray(z['L_full'], dtype=float)
    q = int(0.7 * len(L_dense))
    L_late = float(np.mean(L_dense[q:]))                       # cycle-mean
    LV = np.asarray(z['L_V_opt_train'], dtype=float)
    LV_late = float(np.mean(LV[int(0.7*len(LV)):]))
    _, ymean = _dedupe(z['step'], z['mean_cos2_AGOP'])
    n_log = len(y); mid = n_log // 2
    # plateau window: [onset, T]. Falls back to the whole run if the loss never
    # stops descending (then the "plateau gain" is simply the total gain).
    tp = plateau_onset(L_dense)
    t_lo = float(tp) if tp is not None else float(s[0])
    at = lambda v, t: float(np.interp(t, s, v))
    d_min_plat  = at(y, s[-1])     - at(y, t_lo)
    d_mean_plat = at(ymean, s[-1]) - at(ymean, t_lo)
    # Post-saturation gap: once the features stop improving, how far above the
    # feature-limited optimum does the loss REMAIN? This is the statistic that
    # carries the decoupling for fast-aligning configs (see sat_onset).
    Lp = 0.5 * (L_dense[:-1] + L_dense[1:])            # cycle mean, parity removed
    # NB: LV must be de-duped against its OWN step grid before interpolation --
    # `s` above came from de-duping cos2_min, and a checkpoint resumed mid-run
    # can repeat a step, so the two grids are not the same length.
    s_v, LV_d = _dedupe(z['step'], z['L_V_opt_train'])
    _ok = ~np.isnan(LV_d)
    t_sat = sat_onset(s, ymean)
    gap_sat = float('nan')
    if t_sat == t_sat and len(Lp) > 1 and _ok.sum() >= 2:
        i0 = int(np.clip(t_sat, 0, len(Lp) - 1))
        tt = np.arange(i0, len(Lp), dtype=float)
        gap_sat = float(np.median(Lp[i0:] / np.maximum(np.interp(tt, s_v[_ok], LV_d[_ok]), 1e-12)))
    tau = float('nan')
    if 'tau_V_sq' in z.files:
        tv = np.asarray(z['tau_V_sq'], dtype=float); tv = tv[~np.isnan(tv)]
        if len(tv): tau = float(np.mean(tv[int(0.7*len(tv)):])) if len(tv) > 3 else float(tv[-1])
    return dict(
        rho2=rho2_late(z),
        L_late=L_late, L_V_opt=LV_late,
        ratio=L_late / max(LV_late, 1e-12),
        cm0=float(y[0]), cm_mid=float(y[mid]), cm_end=float(y[-1]),
        mean_end=float(ymean[-1]),
        t_plateau=t_lo,
        t_sat=t_sat,                   # step at which mean cos^2 hits 99% of final
        gap_sat=gap_sat,               # median L/L_V_opt from t_sat to T
        d_mean_plat=d_mean_plat,       # mean cos^2 gained during the plateau
        d_min_plat=d_min_plat,         # cos^2_min gained during the plateau
        t50=t_to(s, y, 0.5), s_band=band_slope(s, y),
        tau_V_sq=tau,
        diverged=L_late > 5.0,
    )

rows = []
for (gid, ta, p, rt, rs, n, eta) in CONFIGS_33:
    for seed in SEEDS:
        ckpt = OUT / f'{ta}_p{p}_rt{rt}_rs{rs}_n{n}_eta{eta}_seed{seed}.npz'
        if not ckpt.exists():
            continue
        m = extract(ckpt)
        if m is None:
            continue
        rows.append(dict(gid=gid, teacher=ta, p=p, r_t=rt, r_s=rs, n=n, eta=eta,
                          M=min(p, rs), seed=seed, ckpt=ckpt.name, **m))
print(f'Extracted {len(rows)} runs ({len(set(r["gid"] for r in rows))}/33 configs present)')
with (OUT/'per_run.csv').open('w', newline='') as f:
    w = csv.DictWriter(f, fieldnames=list(rows[0].keys())); w.writeheader(); w.writerows(rows)
print('Wrote', OUT/'per_run.csv')
```

## 4. The complete 33-config summary table (median ± IQR across seeds)

This is the table the AC asked for. Every config appears; weak/negative cases are flagged, not hidden.

```python
# Single source of truth for verdict labels. Cells 20 and 22 read this instead
# of keeping their own hardcoded copies -- those copies silently went stale
# every time a class was renamed, which is how the breakdown table once
# reported 0 negatives and 8 weak. Red = a real failure to recover the teacher.
VERDICT_COLOR = {
    'clean plateau-with-learning':      'tab:blue',
    'saturates early, gap persists':    'tab:blue',
    'modest plateau (2-3x oracle)':     'tab:blue',
    'no gain in plateau window':        'tab:blue',
    'weakest direction not recovered':  'tab:red',
    'DIVERGES':                         'tab:red',
}
VERDICT_ORDER = list(VERDICT_COLOR)


def med_iqr(vals):
    v = np.array([x for x in vals if x == x])
    if len(v) == 0: return float('nan'), float('nan'), float('nan')
    return float(np.median(v)), float(np.percentile(v, 25)), float(np.percentile(v, 75))

by_cfg = collections.defaultdict(list)
for r in rows: by_cfg[r['gid']].append(r)

summary = []
for (gid, ta, p, rt, rs, n, eta) in CONFIGS_33:
    g = by_cfg.get(gid, [])
    if not g:
        summary.append(dict(gid=gid, teacher=ta, p=p, r_t=rt, r_s=rs, eta=eta,
                            source=PROVENANCE.get(gid, ''), n_seeds=0)); continue
    n_div = sum(1 for r in g if r['diverged'])
    ent = dict(gid=gid, teacher=ta, p=p, r_t=rt, r_s=rs, eta=eta,
               source=PROVENANCE.get(gid, ''),
               M=min(p, rs), n_seeds=len(g), n_diverged=n_div)
    for key in ['rho2', 'ratio', 'd_mean_plat', 'd_min_plat', 'cm_end', 'mean_end',
                'cm0', 't50', 's_band', 'tau_V_sq', 'L_late', 't_plateau',
                't_sat', 'gap_sat']:
        med, q1, q3 = med_iqr([r[key] for r in g])
        ent[f'{key}_med'] = med; ent[f'{key}_q1'] = q1; ent[f'{key}_q3'] = q3
        _v = np.array([r[key] for r in g if r[key] == r[key]])
        ent[f'{key}_sd'] = float(np.std(_v, ddof=1)) if len(_v) > 1 else float('nan')
        ent[f'{key}_n']  = int(len(_v))          # seeds contributing (t50 can be nan)
    # verdict, honest. Note: L_late > 5 alone is NOT divergence -- the
    # cycle-mean is dominated by the high phase of a large-amplitude
    # period-2 cycle. True divergence requires the cycle signature to
    # break down as well (rho2 well below 1).
    # Verdict + the SPECIFIC numbers that triggered it, so every row carries
    # its own justification rather than pointing at a generic rule table.
    _half = len(g) / 2
    if n_div >= _half and ent['rho2_med'] < 0.9:
        ent['verdict'] = 'DIVERGES'
        ent['basis'] = f"{n_div}/{len(g)} seeds Lbar>5 AND rho2={ent['rho2_med']:.3f}<0.9"
    elif ent['cm_end_med'] < 0.3:
        ent['verdict'] = 'weakest direction not recovered'
        ent['basis'] = (f"final cos2min={ent['cm_end_med']:.3f}<0.30 "
                        f"(mean cos2 still rises; tau_V^2={ent['tau_V_sq_med']:.2f})")
    elif ent['d_mean_plat_med'] < 0.02:
        # The plateau window is empty for these: the features finished before
        # the loss stalled, so a ~0 plateau gain is arithmetic, not evidence.
        # Score them on the post-saturation gap instead -- features done, loss
        # still far above what those features permit -- which is decoupling in
        # its strongest form (the loss is not limited by feature quality at all).
        _fast = (ent['t_sat_med'] == ent['t_sat_med']
                 and ent['t_sat_med'] < ent['t_plateau_med'])
        if _fast and ent['gap_sat_med'] >= 3:
            ent['verdict'] = 'saturates early, gap persists'
            ent['basis'] = (f"mean cos2 reached 99% of final at t={ent['t_sat_med']:.0f}, "
                            f"before the plateau formed at t={ent['t_plateau_med']:.0f}, so the "
                            f"plateau gain is ~0 by construction ({ent['d_mean_plat_med']:+.3f}); "
                            f"from t_sat onward cos2={ent['mean_end_med']:.3f} is fixed while the "
                            f"loss stays {ent['gap_sat_med']:.1f}x the feature-limited optimum")
        elif ent['ratio_med'] < 3:
            # Fast aligner whose gap is real but small -- that is exactly the
            # 'modest plateau' class, so route it there rather than inventing a
            # dead-end label for what is just a weaker version of the same thing.
            ent['verdict'] = 'modest plateau (2-3x oracle)'
            ent['basis'] = (f"features saturated at t={ent['t_sat_med']:.0f} (before the plateau), "
                            f"and the loss holds {ent['gap_sat_med']:.1f}x oracle from there on -- "
                            f"decoupling present but not far above the oracle")
        else:
            ent['verdict'] = 'no gain in plateau window'
            ent['basis'] = (f"mean cos2 gained during the plateau = "
                            f"{ent['d_mean_plat_med']:+.3f} < 0.02, and the post-saturation gap "
                            f"is only {ent['gap_sat_med']:.1f}x oracle, so neither signature holds")
    elif ent['ratio_med'] < 3:
        ent['verdict'] = 'modest plateau (2-3x oracle)'
        ent['basis'] = f"L/L_V_opt={ent['ratio_med']:.1f}<3"
    else:
        ent['verdict'] = 'clean plateau-with-learning'
        ent['basis'] = (f"1-rho2={max(1-ent['rho2_med'],1e-10):.0e}, "
                        f"L/L_V_opt={ent['ratio_med']:.1f}, "
                        f"plateau gains dmean={ent['d_mean_plat_med']:+.3f} / "
                        f"dmin={ent['d_min_plat_med']:+.3f}")
    # Cycle amplitude is a property of the LOSS; the verdict describes ALIGNMENT
    # behaviour. Mixing the two put R1 and R11 in one bucket even though R1's
    # features finish at t_sat=630 -- the loss then holds 2500x the
    # feature-limited optimum -- while R11 gains +0.94 *inside* its plateau.
    # Two mechanisms, one label, and it also overrode the d_mean_plat branch
    # so R1 never reached the class that actually describes it. Amplitude now
    # annotates the row; the "large loss is not divergence" point survives in
    # the basis string, and DIVERGES above still catches real blow-ups.
    ent['high_amplitude'] = bool(n_div >= _half)
    if ent['high_amplitude'] and ent['verdict'] != 'DIVERGES':
        ent['basis'] += (f"; large-amplitude cycle: {n_div}/{len(g)} seeds Lbar>5 "
                         f"but 1-rho2={max(1 - ent['rho2_med'], 1e-10):.0e} -- locked, not divergent")
    summary.append(ent)

hdr = f"{'id':>4} {'act':>5} {'p':>4} {'rt':>3} {'rs':>3} {'eta':>5} {'sds':>3} {'div':>3} {'1-rho2':>9} {'ratio':>8} {'t_plat':>7} {'t_sat':>6} {'gap_sat':>8} {'dmean_pl':>9} {'dmin_pl':>8} {'cm_end':>16} {'t50':>6} {'tauV2':>6}  verdict"
print(hdr); print('-'*len(hdr))
for e in summary:
    if e.get('n_seeds', 0) == 0:
        print(f"{e['gid']:>4} {e['teacher']:>5} {e.get('p',100):>4} {e['r_t']:>3} {e['r_s']:>3} {e['eta']:>5}  "
              f"(NO RUNS YET -- {PROVENANCE.get(e['gid'],'')})"); continue
    print(f"{e['gid']:>4} {e['teacher']:>5} {e.get('p',100):>4} {e['r_t']:>3} {e['r_s']:>3} {e['eta']:>5} {e['n_seeds']:>3} {e['n_diverged']:>3} "
          f"{max(1-e['rho2_med'],1e-10):>9.1e} {e['ratio_med']:>8.1f} {e['t_plateau_med']:>7.0f} {e['t_sat_med']:>6.0f} {e['gap_sat_med']:>8.1f} {e['d_mean_plat_med']:>+9.3f} {e['d_min_plat_med']:>+8.3f} {e['cm_end_med']:>8.3f}+/-{e['cm_end_sd']:<7.3f} "
          f"{e['t50_med']:>6.0f} {e['tau_V_sq_med']:>6.3f}  {e['verdict']}")

with (OUT/'summary_33.csv').open('w', newline='') as f:
    keys = sorted({k for e in summary for k in e})
    w = csv.DictWriter(f, fieldnames=keys); w.writeheader(); w.writerows(summary)
print('\nWrote', OUT/'summary_33.csv')
```

## 5b. Trajectories — all 33, and by outcome class


```python
# ---------------------------------------------------------------------------
# Per-configuration trajectories: loss, mean cos^2, cos^2_min -- one
# small-multiple grid per quantity, all 33 configurations, every seed drawn.
#
# ENCODING, stated so nothing has to be inferred from thickness:
#     faint thin  = one seed (cycle mean)      -> seed dispersion
#     pale ribbon = MEDIAN raw per-step loss   -> the period-2 oscillation
#     heavy       = MEDIAN cycle mean          -> the trend
#     dashed      = head-fit oracle L_V_opt
# The two medians are the same 50 runs summarised two ways, in two shades of
# the same hue, and both are named in the legend. Previously only the raw loss
# was drawn (thin, per seed) so the panel's thickness WAS the oscillation --
# ~171% of the cycle mean on R3 -- while the alignment panels' thickness was
# seed spread, ~1.6% on the same run. Same visual, two meanings, two orders of
# magnitude apart, which is how an oscillation ribbon gets read as an error
# bar. The elementwise median across seeds preserves the alternation (all seeds
# share step parity; measured rho2 = 1.000), so the ribbon is a real summary
# and not an artefact of averaging.
# ---------------------------------------------------------------------------
import matplotlib.colors as _mcol

def _pale(c, f=0.62):
    """Same hue, blended toward white -- keeps the verdict colour readable
    while making clear the ribbon and the trend are the same runs."""
    r, g, b = _mcol.to_rgb(c)
    return (r + (1-r)*f, g + (1-g)*f, b + (1-b)*f)

NEG_VERDICT = 'weakest direction not recovered'

def vcolor(verdict):
    return VERDICT_COLOR.get(str(verdict), 'tab:blue')

_present = {e.get('verdict', '') for e in summary if e.get('n_seeds', 0) > 0}
VCOL = {v: VERDICT_COLOR[v] for v in VERDICT_ORDER if v in _present}
print('by-verdict panels:', list(VCOL))
vmap = {e['gid']: e.get('verdict', '') for e in summary}

_TRACE_CACHE = {}

def traces(gid):
    """Cached: the three grids each ask for all 33 configs, so without this the
    checkpoints are re-read three times over."""
    if gid in _TRACE_CACHE:
        return _TRACE_CACHE[gid]
    out = []
    _, ta, p, rt, rs, n, eta = next(c for c in CONFIGS_33 if c[0] == gid)
    for sd in SEEDS:
        ck = OUT / f'{ta}_p{p}_rt{rt}_rs{rs}_n{n}_eta{eta}_seed{sd}.npz'
        if not ck.exists(): continue
        try:
            z = np.load(ck, allow_pickle=True); _ = [z[k] for k in z.files]
        except Exception: continue
        s, cm = _dedupe(z['step'], z['cos2_min_AGOP'])
        _, mc = _dedupe(z['step'], z['mean_cos2_AGOP'])
        sv, LV = _dedupe(z['step'], z['L_V_opt_train'])
        L = np.asarray(z['L_full_dense'], float)         # raw per-step loss
        out.append(dict(s=s, cm=cm, mc=mc, L=L,
                        Lb=0.5*(L[:-1]+L[1:]), sv=sv, LV=LV))
    _TRACE_CACHE[gid] = out
    return out

OSC_WIN = 60          # steps shown in the oscillation inset

def grid_figure(kind, fname, ylab, title):
    nrow, ncol = 6, 6
    fig, axes = plt.subplots(nrow, ncol, figsize=(3.1*ncol, 2.4*nrow), sharex=True)
    for k, (gid, ta, p, rt, rs, n, eta) in enumerate(CONFIGS_33):
        ax = axes[k // ncol][k % ncol]
        T = traces(gid); col = vcolor(vmap.get(gid, ''))
        if T:
            if kind == 'loss':
                nL = min(len(t['Lb']) for t in T)
                nR = min(len(t['L']) for t in T)
                x = np.arange(nL)
                for t in T:                       # faint thin = ONE SEED
                    ax.semilogy(x, t['Lb'][:nL], color=col, lw=0.45, alpha=0.22,
                                zorder=2)
                # median of the RAW per-step loss: the period-2 cycle, summarised
                # over seeds rather than shown for one. Drawn behind the trend in
                # a pale tint of the same hue.
                ax.semilogy(np.arange(nR),
                            np.median(np.vstack([t['L'][:nR] for t in T]), axis=0),
                            color=_pale(col), lw=0.6, alpha=0.9, zorder=1)
                ax.semilogy(x, np.median(np.vstack([t['Lb'][:nL] for t in T]), axis=0),
                            color=col, lw=1.8, zorder=3)
                ax.semilogy(T[0]['sv'],
                            np.median(np.vstack([np.interp(T[0]['sv'], t['sv'], t['LV'])
                                                 for t in T]), axis=0),
                            color='k', lw=1.0, ls='--', zorder=4)
                # Inset: the period-2 oscillation itself, one seed, late window.
                # This is the object whose amplitude used to be the panel's
                # thickness; giving it its own axes makes the main panel's
                # thickness mean seeds and nothing else.
                raw = T[0]['L']
                a, b = max(0, len(raw) - OSC_WIN - 10), max(1, len(raw) - 10)
                ins = ax.inset_axes([0.55, 0.56, 0.42, 0.38])
                ins.plot(np.arange(a, b), raw[a:b], color=col, lw=0.7)
                ins.set_xticks([]); ins.set_yticks([])
                for sp in ins.spines.values():
                    sp.set_linewidth(0.5); sp.set_color('0.5')
                ins.set_title(f'raw, {b-a} steps', fontsize=5.2, pad=1.5, color='0.35')
            else:
                key = 'mc' if kind == 'mean' else 'cm'
                for t in T:                       # thin = ONE SEED
                    ax.plot(t['s'], t[key], color=col, lw=0.6, alpha=0.30)
                grid = min((t['s'] for t in T), key=len)
                ax.plot(grid, np.median(np.vstack([np.interp(grid, t['s'], t[key])
                                                   for t in T]), axis=0), color=col, lw=1.9)
                ax.set_ylim(-0.03, 1.03)
            ax.text(.03, .95, f"{gid}  {ta} $r_t$={rt} $r_s$={rs} $\\eta$={eta}  n={len(T)}",
                    transform=ax.transAxes, fontsize=6.5, va='top')
        ax.grid(alpha=.25); ax.tick_params(labelsize=7)
    for k in range(len(CONFIGS_33), nrow*ncol): axes[k//ncol][k%ncol].axis('off')
    for a in axes[-1]: a.set_xlabel('step', fontsize=8)
    for r in range(nrow): axes[r][0].set_ylabel(ylab, fontsize=8)
    hs = [plt.Line2D([], [], color='tab:blue', lw=2,
                     label=f'phenomenon present ({sum(1 for v in vmap.values() if not str(v).startswith("weakest"))}/33)'),
          plt.Line2D([], [], color='tab:red', lw=2,
                     label=f'weakest direction not recovered '
                           f'({sum(1 for v in vmap.values() if str(v).startswith("weakest"))}/33)')]
    if kind == 'loss':
        hs += [plt.Line2D([], [], color='0.55', lw=0.6,
                          label='median raw loss (period-2 cycle)'),
               plt.Line2D([], [], color='0.25', lw=1.8, label='median cycle mean'),
               plt.Line2D([], [], color='0.55', lw=0.45, alpha=0.5,
                          label='individual seeds (cycle mean)'),
               plt.Line2D([], [], color='k', lw=1, ls='--', label='$L_{V,opt}$ oracle')]
    else:
        hs += [plt.Line2D([], [], color='0.25', lw=1.9, label='seed median'),
               plt.Line2D([], [], color='0.55', lw=0.6, alpha=0.5,
                          label='individual seeds')]
    fig.legend(handles=hs, loc='lower center', ncol=min(len(hs), 4), fontsize=8, frameon=False,
               bbox_to_anchor=(0.5, -0.012))
    fig.suptitle(title, fontsize=12)
    fig.tight_layout(rect=[0, 0.014, 1, 0.985])
    fig.savefig(OUT/fname, dpi=130, bbox_inches='tight'); plt.show()

grid_figure('loss', 'all33_loss.png', 'cycle-mean loss',
            'All 33 configurations: cycle-mean loss $\\bar L_t=(L_t+L_{t+1})/2$. '
            'Pale ribbon = median raw loss (the period-2 cycle), heavy = median '
            'cycle mean, faint = individual seeds, dashed = head-fit oracle. '
            'Inset: the same oscillation resolved over 60 steps (one seed).')
grid_figure('mean', 'all33_mean_cos2.png', r'mean $\cos^2$',
            r'All 33 configurations: mean $\cos^2$ (thin = one seed, heavy = seed median)')
grid_figure('min', 'all33_min_cos2.png', r'$\cos^2_{\min}$',
            r'All 33 configurations: $\cos^2_{\min}$, the weakest teacher direction '
            r'(thin = one seed, heavy = seed median)')
```

```python

# ---------------------------------------------------------------------------
# The outcome classes, one column each. Same encoding rule as the grids above:
# shaded = seed dispersion (IQR over seeds), line = seed median. Nothing here
# is the period-2 oscillation -- row 1 plots the CYCLE MEAN over the oracle, so
# the alternation is already divided out.
# ---------------------------------------------------------------------------
classes = [k for k in VCOL if any(v == k for v in vmap.values())]
fig, axes = plt.subplots(3, len(classes), figsize=(4.3*len(classes), 10), squeeze=False)
cmap = plt.get_cmap('viridis')

def _q(M):
    return np.percentile(M, [25, 50, 75], axis=0)

for col, cls in enumerate(classes):
    gids = [g for g, v in vmap.items() if v == cls]
    for j, gid in enumerate(gids):
        T = traces(gid)
        if not T: continue
        c = cmap(j / max(len(gids)-1, 1))
        grid = min((t['s'] for t in T), key=len)
        # t['cm'] is cos2_min_AGOP; t['mc'] is mean_cos2_AGOP.
        MC = np.vstack([np.interp(grid, t['s'], t['mc']) for t in T])
        CN = np.vstack([np.interp(grid, t['s'], t['cm']) for t in T])
        # traces() returns LV on the SPARSE log grid (t['sv']) while Lb is dense
        # per-step -- interpolate before dividing (7999 vs 401).
        nL = min(len(t['Lb']) for t in T)
        RAT = np.vstack([t['Lb'][:nL] / np.maximum(
            np.interp(np.arange(nL), t['sv'], t['LV']), 1e-12) for t in T])
        for a, X, xs, logy in ((axes[0][col], RAT, np.arange(nL), True),
                               (axes[1][col], MC, grid, False),
                               (axes[2][col], CN, grid, False)):
            q1, med, q3 = _q(X)
            a.fill_between(xs, q1, q3, color=c, alpha=0.15, lw=0)
            (a.semilogy if logy else a.plot)(xs, med, color=c, lw=1.4,
                                             label=gid if logy else None)
    axes[0][col].axhline(1.0, color='k', lw=1, ls='--')
    axes[0][col].set_title(cls.split(' (')[0] + f'  (n={len(gids)})', fontsize=10)
    axes[0][col].legend(fontsize=6.5, ncol=2)
    for a in axes[:, col]: a.grid(alpha=.3)
    axes[1][col].set_ylim(-0.02, 1.02)
    axes[2][col].set_ylim(-0.02, 1.02)
    axes[2][col].set_xlabel('step')
axes[0][0].set_ylabel(r'cycle-mean $\bar L\ /\ L_{V,\mathrm{opt}}$  (1 = oracle)')
axes[1][0].set_ylabel(r'mean $\cos^2$')
axes[2][0].set_ylabel(r'$\cos^2_{\min}$  (weakest direction)')
fig.suptitle('The five outcome classes: line = seed median, band = IQR over seeds',
             fontsize=12)
fig.tight_layout(); fig.savefig(OUT/'all33_by_verdict.png', dpi=130, bbox_inches='tight'); plt.show()
```
