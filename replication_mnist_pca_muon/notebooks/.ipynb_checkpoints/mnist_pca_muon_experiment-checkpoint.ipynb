# Parametric PCA-MNIST exact-polar Muon experiment

This notebook can run a quick smoke test or reproduce the full experiment. The main parameter cell exposes `n_train`, PCA dimension and whitening, teacher rank, student width, fixed learning rate, steps, seed lists, and analysis windows.

```python
from pathlib import Path
import copy
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "mnist_pca_muon.py").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

import mnist_pca_muon as mpm
import pandas as pd
from IPython.display import Image, display

print("Bundle root:", ROOT)
```

```python
from pathlib import Path

# ------------------------------------------------------------
# Exact reported experiment
# ------------------------------------------------------------
PRESET = "reproduction"
PROBLEM_MODE = "frozen"

RUN_CONFIRMATION = True
FORCE_REBUILD_PROBLEM = False
FORCE_RERUN = True

# The exact frozen problem included in the bundle.
FROZEN_PROBLEM = (
    ROOT / "data" / "fixed_problem_reproduction.npz"
).resolve()

assert FROZEN_PROBLEM.is_file(), (
    f"Frozen problem is missing: {FROZEN_PROBLEM}"
)

config = mpm.load_config(
    ROOT / "configs" / "reproduction.json"
)

# Use a new output directory so no smoke-test results are reused.
config["output"]["output_root"] = str(
    (ROOT / "notebook_outputs").resolve()
)
config["output"]["experiment_name"] = (
    "notebook_reproduction_exact"
)

# ------------------------------------------------------------
# Problem-defining parameters
# ------------------------------------------------------------
config["data"]["n_train"] = 5000
config["data"]["pca_dim"] = 100
config["data"]["whiten"] = True
config["data"]["problem_seed"] = 2026
config["data"]["allowed_digits"] = None

config["teacher"]["rank"] = 4
config["teacher"]["seed"] = 5
config["teacher"]["head_value"] = 1.0

# ------------------------------------------------------------
# Student/training parameters
# ------------------------------------------------------------
config["student"]["width"] = 50
config["student"]["head_value"] = None
config["student"]["learning_rate"] = 0.80 #.6 earlier
config["student"]["steps"] = 1500
config["student"]["log_every"] = 10
config["student"]["log_both_phases"] = True
config["student"]["threads"] = 1
config["student"]["checkpoint_steps"] = [
    0,
    60,
    240,
    400,
    700,
    1000,
    1200,
    1490,
]

# ------------------------------------------------------------
# Exact seed sets
# ------------------------------------------------------------
config["seeds"]["development"] = [
    5, 11, 23, 37, 41
]

config["seeds"]["confirmation"] = [
    61, 67, 71, 79, 83, 89, 97, 101, 103, 107,
    109, 113, 127, 131, 137, 139, 149, 151, 157, 163,
    167, 173, 179, 181, 191, 193, 197, 199, 211, 223,
]

# ------------------------------------------------------------
# Analysis
# ------------------------------------------------------------
config["analysis"]["primary_window"] = [60, 240]
config["analysis"]["bootstrap_repetitions"] = 20000
config["analysis"]["plot_xmax"] = 1490

mpm.validate_config(config)

print("Preset:", PRESET)
print("Problem mode:", PROBLEM_MODE)
print("Frozen problem:", FROZEN_PROBLEM)
print("Confirmation seeds:", len(config["seeds"]["confirmation"]))

config
```

## 2. Prepare the fixed problem

```python
# problem_path = mpm.prepare_problem(config, force=FORCE_REBUILD_PROBLEM)
import shutil

experiment_dir = mpm.experiment_dir(config)
experiment_dir.mkdir(parents=True, exist_ok=True)

problem_path = (
    experiment_dir
    / config["output"]["problem_filename"]
)

# Copy the exact fixed problem into this experiment directory.
shutil.copy2(
    FROZEN_PROBLEM,
    problem_path,
)

problem = mpm.load_problem(problem_path)

print("Problem:", problem_path)
print("X:", problem["X"].shape)
print("Teacher basis:", problem["U_teacher"].shape)

# Fail immediately if smoke settings are accidentally still active.
assert problem["X"].shape == (5000, 100)
assert problem["U_teacher"].shape == (100, 4)

assert config["student"]["width"] == 50
assert config["student"]["learning_rate"] == 0.80 # .6 earlier
assert config["student"]["steps"] == 1500
assert config["analysis"]["primary_window"] == [60, 240]
assert len(config["seeds"]["confirmation"]) == 30


# problem = mpm.load_problem(problem_path)
# print("Problem:", problem_path)
# print("X:", problem["X"].shape)
# print("Teacher basis:", problem["U_teacher"].shape)
# pd.read_csv(problem_path.with_name("problem_summary.csv"))
```

## 3. Train students

The full-batch update is $W_{t+1}=W_t-\eta\,\mathrm{Polar}(\nabla_W L(W_t))$. The head is fixed.

```python
groups = ("development", "confirmation") if RUN_CONFIRMATION else ("development",)
run_outputs = mpm.run_experiment(
    config,
    groups=groups,
    rebuild_problem=False,
    force_runs=FORCE_RERUN,
)
run_outputs
```

## 4. Analyze the loss cycle and both alignment phases

```python
analysis_outputs = mpm.analyze_experiment(config, groups=groups)
summary = pd.read_csv(analysis_outputs["summary"])
summary
```

```python
analysis_dir = analysis_outputs["analysis_dir"]
for group in groups:
    for filename in [
        f"{group}_loss_cycle.png",
        f"{group}_mean_alignment_two_phases.png",
        f"{group}_min_alignment_two_phases.png",
        f"{group}_mean_alignment_phase_average.png",
        f"{group}_min_alignment_phase_average.png",
        f"{group}_two_step_residual.png",
    ]:
        path = analysis_dir / filename
        if path.exists():
            display(Image(filename=str(path)))
```

## 5. Optional fixed-learning-rate sweep

```python
RUN_ETA_SWEEP = False
ETA_GRID = [0.4, 0.5, 0.6, 0.7, 0.8]

if RUN_ETA_SWEEP:
    sweep_config = copy.deepcopy(config)
    sweep_config["output"]["experiment_name"] = f"notebook_{PRESET}_sweep"
    sweep_table = mpm.sweep_learning_rates(
        sweep_config,
        ETA_GRID,
        group="development",
        force=FORCE_RERUN,
    )
    display(sweep_table)
else:
    print("Set RUN_ETA_SWEEP=True to run the sweep.")
```

The teacher basis is drawn independently in PCA coordinates. Whitening controls covariance but does not make the empirical MNIST distribution Gaussian. The task still uses teacher-generated targets.
